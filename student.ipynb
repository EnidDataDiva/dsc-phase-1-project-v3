## Data Preparation

### Dropping irrelevant columns & rows

```python
# Dropping potentially irrelevant columns
df = df.drop(columns=[
    'Airport.Code', 
    'Airport.Name', 
    'Schedule', 
    'Amateur.Built',  
    'FAR.Description', 
    'Registration.Number',
    'Longitude',
    'Latitude', 
    'Air.carrier', 
    'Publication.Date'
])
```

```python
# Clean column names
df.columns = (df.columns
               .str.strip()
               .str.replace('.', '_')
               .str.replace(' ', '_')
               .str.lower())
```

```python
# Iterate through each column in the DataFrame that has an object data type (typically strings)
for col in df.select_dtypes(include=['object']).columns:
    # Clean the current column by applying several string operations
    df[col] = (df[col]  
                .str.strip()  
                .str.replace('.', '_')  
                .str.replace(' ', '_') 
                .str.lower())
```

```python
# Dropping aircrafts that arent airplanes because we only want airplanes
df = df[df['aircraft_category'] == 'airplane']

df['aircraft_category'].value_counts()
```

```python
#Identify the missing values
df.isnull().sum()
```

```python
# Impute categorical columns with mode
df['location'].fillna(df['location'].mode()[0], inplace=True)
df['country'].fillna(df['country'].mode()[0], inplace=True)
```

```python
# Impute numerical columns with median
df['total_minor_injuries'].fillna(df['total_minor_injuries'].median(), inplace=True)
df['total_uninjured'].fillna(df['total_uninjured'].median(), inplace=True)
```

```python
#Display the event date column
print(df['event_date'].unique())
```

```python
# Convert 'event_date' to datetime format
df['event_date'] = pd.to_datetime(df['event_date'], errors='coerce')
```

```python
# Create a new column for the year
df['event_year'] = df['event_date'].dt.year
```

```python
# Create a cleaned version of df
df_clean = df.copy() 
```

```python
# Check if there are any more null values in our data set
df_clean.isnull().any().sum()
```

```python
# Check which specific columns have missing values
missing_columns = df_clean.columns[df_clean.isnull().any()]
print(missing_columns)
```

```python
# Impute 'injury_severity' using the mode
injury_mode = df['injury_severity'].mode()[0]
df['injury_severity'].fillna(injury_mode, inplace=True)

# Impute 'total_fatal_injuries' and 'total_serious_injuries'
# If an entry is missing, replace it with 0 (assuming no injuries)
df['total_fatal_injuries'].fillna(0, inplace=True)
df['total_serious_injuries'].fillna(0, inplace=True)

# Check the results of imputation
print(df[['injury_severity', 'total_fatal_injuries', 'total_serious_injuries']].isnull().sum())

# Convert injury_severity to categorical if needed
df['injury_severity'] = pd.Categorical(df['injury_severity'])

# Display unique values in the 'injury severity' column
unique_injury_severity = df['injury_severity'].unique()
print(unique_injury_severity)

# Function to clean the Injury Severity values
def clean_injury_severity(severity):
    if 'fatal' in severity.lower():
        return 'Fatal'
    elif 'serious' in severity.lower():
        return 'Serious'
    elif 'minor' in severity.lower():
        return 'Minor'
    else:
        return 'Non-Fatal'

# Apply the function to the Injury Severity column
df['injury_severity'] = df['injury_severity'].apply(clean_injury_severity)

# Display unique values to check the cleaned column
unique_injury_severity = df['injury_severity'].unique()
print(unique_injury_severity)
```

```python
# Check the distribution of weather_condition
weather_distribution = df['weather_condition'].value_counts()
print(weather_distribution)

# Impute 'weather_condition' using the most frequent value
most_frequent_weather = df['weather_condition'].mode()[0]
df['weather_condition'].fillna(most_frequent_weather, inplace=True)

# Check the results of imputation
print(df['weather_condition'].isnull().sum())
```

```python
# drop nulls in Model and Model column
df_columns = df.dropna(subset = ['model', 'make'])
df_columns
```

```python
# Check any duplicate entries
df_clean.duplicated().any()
```

```python
# Identify duplicate entries
duplicates = df_clean[df_clean.duplicated()]
print("Duplicate entries:")
print(duplicates) 

# Drop duplicates
df_clean.drop_duplicates(inplace=True)

# Recheck for duplicates
print("Any duplicates remaining?", df_clean.duplicated().any())
```

### Analysis of Aviation Accidents Over the Years


```python
# Extract the year from event_date
df['year'] = pd.to_datetime(df['event_date']).dt.year

# Plotting the histogram
plt.figure(figsize=(10, 6))
sns.histplot(df['year'], bins=range(1962, 2024), kde=True)
plt.title('Distribution of Aviation Accidents Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Accidents')
plt.grid()
plt.show()
```

The histogram illustrates a decreasing trend in the number of aviation accidents over the years, suggesting improvements in aviation safety and regulatory measures. This decline could be attributed to advancements in technology, better training for pilots, and stricter safety regulations. Further analysis is needed to understand the specific factors contributing to this positive trend.

### Aircraft Manufacturer by Number of Accidents

```python
# Grouping by aircraft manufacturer
manufacturer_counts = df['make'].value_counts().head(10)

# Plotting the bar graph
plt.figure(figsize=(12, 6))
manufacturer_counts.plot(kind='bar')
plt.title('Top 10 Aircraft Manufacturers by Number of Accidents')
plt.xlabel('Aircraft Manufacturer')
plt.ylabel('Number of Accidents')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

The bar graph illustrates the number of accidents associated with various aircraft manufacturers, ranked from highest to lowest. The analysis reveals the following key findings:

1. **Cessna** is the manufacturer with the highest number of reported accidents, indicating a potential need for further investigation into the safety measures and operational practices associated with this brand.

2. **Piper** and **Beech** follow as the second and third most frequently involved manufacturers in accidents. This highlights a significant presence in aviation incidents that warrants attention.

3. Other notable manufacturers, such as **Boeing**, **Bell**, **Mooney**, **Robinson**, **Grumman**, **Belanca**, and **Hughes**, also appear in the data, though their accident counts are lower compared to the top three.

4. The data suggests that certain manufacturers have a higher incidence of accidents, which could be reflective of their prevalence in the aviation market or other factors that might influence safety.

### Total Injuries

```python
# Summing the total number of injuries by severity level
injury_counts = df[['total_fatal_injuries', 'total_serious_injuries', 'total_minor_injuries']].sum()

# Setting up the figure size for better visibility
plt.figure(figsize=(10, 6))

# Plotting a stacked bar chart to display the distribution of total injuries
injury_counts.plot(kind='bar', stacked=True, color=['red', 'orange', 'yellow'])

# Adding a title to the chart for clarity
plt.title('Total Injuries Distribution')

# Labeling the x-axis to indicate the type of injury severity
plt.xlabel('Injury Severity')

# Labeling the y-axis to show the total number of injuries
plt.ylabel('Total Number of Injuries')

# Rotating x-ticks for better readability
plt.xticks(rotation=0)

# Adding grid lines along the y-axis for easier interpretation of values
plt.grid(axis='y')

# Displaying the plot
plt.show()
```

The bar chart illustrates the distribution of total injuries categorized by severity:

- **Fatal Injuries**: The highest bar represents the total number of fatal injuries, indicating that this is the most severe and critical outcome in aviation accidents. 
- **Minor Injuries**: The second highest bar shows the total number of minor injuries, suggesting that while these incidents are less severe than fatalities, they still represent a significant concern.
- **Serious Injuries**: The middle bar represents serious injuries, which are the lowest in total among the three categories. This indicates that serious injuries occur less frequently than both fatal and minor injuries in the dataset.
